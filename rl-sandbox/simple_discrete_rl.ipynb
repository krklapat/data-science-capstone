```python
import random
import heapq
```

```python
#memory construction
memory = {}
for x1 in range(1,11):
    for x2 in range(1,11):
        memory[x1,x2] = {}
        for y in ("First", "Last", "Equal"):
            memory[x1,x2][y] = 1
```

```python
memory
```

```python
class Actor:
    def __init__(self, memory):
        self.memory = memory

    def choose_action(self, x1, x2):
        data = memory[x1,x2]
        sorted_keys = sorted(data, key = data.get)

        total = sum(data.values())
        i = 0
        key = sorted_keys[i]
        confidence = data[key] / total
        action = key
        while(random.uniform(0,1) > confidence and i < len(data)-1):
            i += 1
            key = sorted_keys[i]
            confidence = data[key] / total
            action = key
        
        return action, confidence

    def update_reward(self,x1,x2,action,reward):
        memory[x1,x2][action] += reward
```

```python
#train
def run(memory, test):
    if not test:
        n = 10000
    else:
        n = 1
    actor = Actor(memory)
    for i in range(n):
        x1 = random.randint(1,10)
        x2 = random.randint(1,10)
        action, confidence = actor.choose_action(x1,x2)
        
        if x1 > x2 and action == "First" or x1 < x2 and action == "Last" or x1 == x2 and action == "Equal":
            reward = 1
        else:
            reward = 0
        actor.update_reward(x1,x2,action,reward)
        if test:
            print(x1, x2, action, confidence, reward)
            print(actor.memory)


```

```python
run(memory, test=False)
```

```python
run(memory, test=True)
```
