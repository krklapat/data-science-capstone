**Optimizatoin Algorithm**

```python
import pandas as pd
```

```python
energy = pd.read_parquet('../Data/cleaned_data/dam.parquet')
energy.head()
```

```python
avg_lmp = energy['lmp'].mean()
avg_lmp
```

```python
#standard average
battery = 0
spent = 0
earned = 0
buys = 0
sells = 0

for price in energy['lmp']:
        if price > avg_lmp and battery == 1:
            battery = 0
            earned += price
            sells += 1

        elif price < avg_lmp and battery == 0:
            battery = 1
            spent += price  
            buys += 1   
profit = earned - spent

print('Total amount spent:', spent)
print('Total amount earned:', earned)
print('Total profit:', profit)
print('Total number of times bought:', buys)
print('Total number of times sold:', sells)
```

```python
energy['rolling_avg'] = energy['lmp'].shift(1).rolling(window=288).mean()
```

```python
#rolling averge
battery = 0
spent = 0
earned = 0
buys = 0
sells = 0

for price, avg in zip(energy['lmp'],energy['rolling_avg']):
        if price > avg and battery == 1:
                battery = 0
                earned += price
                sells += 1
    
        elif price < avg and battery == 0:
                battery = 1
                spent += price  
                buys += 1  
profit = earned - spent

print('Total amount spent:', spent)
print('Total amount earned:', earned)
print('Total profit:', profit)
print('Total number of times bought:', buys)
print('Total number of times sold:', sells)               
```
