```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_parquet("..\\data-pull\\combined_data\\rtm_data_2023_2026.parquet")
```

```python
rtm = data.to_dict(orient='list')
```

```python
rtm["lmp"][1]
```

```python
rtm["predicted"] = [-9999] * len(rtm["lmp"])
rtm["predicted"]
```

```python
rtm
```

```python
for i in range(len(rtm["lmp"])):
    if i >= 24:
        rtm["predicted"][i] = np.mean(rtm["lmp"][i-24:i])
    else:
        rtm["predicted"][i] = np.mean(rtm["lmp"][0:i])
```

```python
rtm_df = pd.DataFrame(rtm)
mse = np.mean((rtm_df["lmp"] - rtm_df["predicted"]) ** 2)
print(mse)
```
