```python
import os
import pandas as pd
from gridstatusio import GridStatusClient
```

### Constants

To get API Key go to gridstatus website > Settings > API > generate API Key

```python
# Initialize client (Set API key in environment or pass directly)
#os.environ["GRIDSTATUS_API_KEY"] = "API_KEY"  # Replace with your actual API key
client = GridStatusClient()

hub_location = "MINN.HUB"
start_date = "2025-01-01"
end_date = "2025-01-02"
```

### 5-min Real Time LMP

```python
print(f"--- 1. Pulling 5-Minute Real-Time LMP for {hub_location} ---")
df_rtm = client.get_dataset(
    dataset="miso_lmp_real_time_5_min",
    start=start_date,
    end=end_date,
    filter_column="location",
    filter_value=hub_location,
)
print("Real-Time Sample Rows:")
print(df_rtm[["interval_start_utc", "location", "lmp", "energy", "congestion"]].head(3))
```

### 1-Hour Day-Ahead LMP

```python
print(f"--- 2. Pulling 1-Hour Day-Ahead LMP for {hub_location} ---")
df_dam = client.get_dataset(
    dataset="miso_lmp_day_ahead_hourly",
    start=start_date,
    end=end_date,
    filter_column="location",
    filter_value=hub_location,
)
print("Day-Ahead Sample Rows:")
print(df_dam[["interval_start_utc", "location", "lmp"]].head(3))
```

### Load Forecast

```python
print("--- 3. Pulling MISO Load Forecast ---")
df_load = client.get_dataset(
    dataset="miso_load_forecast",
    start=start_date,
    end=end_date,
)
print("Load Forecast Sample Rows:")
print(df_load[["interval_start_utc", "load_forecast"]].head(3))
```

### Solar Forecast

```python
print("--- 4. Pulling MISO Solar Forecast ---")
df_solar = client.get_dataset(
    dataset="miso_solar_forecast_hourly",
    start=start_date,
    end=end_date,
)
print("Solar Forecast Sample Rows:")
print(df_solar.head(3))
```

### Wind Forecast

```python
print("--- 5. Pulling MISO Wind Forecast ---")
df_wind = client.get_dataset(
    dataset="miso_wind_forecast_hourly",
    start=start_date,
    end=end_date,
)
print("Wind Forecast Sample Rows:")
print(df_wind.head(3))
```

Solar and wind forecasts look to be published about a week out and then updated every day as they get closer. Depending on how far out we want to look, we can filter this to make pull request smaller.

### Generation Outage Forecast

```python
print("--- 6. Pulling MISO Outage Forecast ---")
df_outage = client.get_dataset(
    dataset="miso_generation_outages_forecast",
    start=start_date,
    end=end_date,
)
print("Outage Forecast Sample Rows:")
print(df_outage.head(3))
```

Same for outage forecasts, except they are also broken up by region.

## Save data locally

```python
os.makedirs("data_sample", exist_ok=True)
df_rtm.to_parquet("data_sample/miso_rtm_north_hub_sample.parquet")
df_dam.to_parquet("data_sample/miso_dam_north_hub_sample.parquet")
df_load.to_parquet("data_sample/miso_load_forecast_sample.parquet")
df_solar.to_parquet("data_sample/miso_solar_forecast_sample.parquet")
df_wind.to_parquet("data_sample/miso_wind_forecast_sample.parquet")
df_outage.to_parquet("data_sample/miso_outage_forecast_sample.parquet")
print("Sample data saved!")
```
