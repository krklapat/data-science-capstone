# Pulling Grid Status Data

```python
import os
import pandas as pd
from gridstatusio import GridStatusClient
```

```python
# Initialize client (Set API key in environment or pass directly)
# To get API Key go to gridstatus website > Settings > API > generate API Key
#os.environ["GRIDSTATUS_API_KEY"] = "API_KEY"  # Replace with your actual API key
client = GridStatusClient()
```

```python
# Define parameters for data pull
hub_location = "MINN.HUB"
start_date = "2016-01-01"
end_date = "2017-01-01"
```

## 5 minute real time LMP

```python
# Pull Data from GridStatus API
df_rtm = client.get_dataset(
    dataset="miso_lmp_real_time_5_min",
    start=start_date,
    end=end_date,
    filter_column="location",
    filter_value=hub_location,
)
print("Real-Time Sample Rows:")
print(df_rtm.head(10))
```

```python
df_rtm.info()
```

```python
# Save the data to a parquet file
os.makedirs("raw_data", exist_ok=True)
df_rtm.set_index("interval_start_utc", inplace=True)
df_rtm.to_parquet(f"raw_data/rtm_data_{start_date}_{end_date}.parquet", index=True)
```

## 1 Hour Day Ahead LMP

```python
df_dam = client.get_dataset(
    dataset="miso_lmp_day_ahead_hourly",
    start=start_date,
    end=end_date,
    filter_column="location",
    filter_value=hub_location,
)
print("Day-Ahead Sample Rows:")
print(df_dam.head(10))
```

```python
# Save the day-ahead data to a parquet file
os.makedirs("raw_data", exist_ok=True)
df_dam.set_index("interval_start_utc", inplace=True)
df_dam.to_parquet(f"raw_data/dam_data_{start_date}_{end_date}.parquet", index=True)
```

## Load Forecast

```python
df_load = client.get_dataset(
    dataset="miso_load_forecast",
    start=start_date,
    end=end_date,
)
print("Load Forecast Sample Rows:")
print(df_load.head(10))
```

```python
# Save the Load Forecast data to a parquet file
os.makedirs("raw_data", exist_ok=True)
df_load.set_index("interval_start_utc", inplace=True)
df_load.to_parquet(f"raw_data/load_data_{start_date}_{end_date}.parquet", index=True)
```

## Solar Forecast

```python
df_solar = client.get_dataset(
    dataset="miso_solar_forecast_hourly",
    start=start_date,
    end=end_date,
)
print("Solar Forecast Sample Rows:")
print(df_solar.head(10))
```

```python
# Save the Solar Forecast data to a parquet file
os.makedirs("raw_data", exist_ok=True)
# Don't index by interval_start_utc for solar data, as each interval is published 6 times (one for each forecast horizon)
df_solar.to_parquet(f"raw_data/solar_data_{start_date}_{end_date}.parquet", index=False)
```
